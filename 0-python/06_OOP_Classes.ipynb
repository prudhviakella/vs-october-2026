A **class** groups related data and the functions that work with that data. Think of a class as a blueprint and an object as one item made from that blueprint.

## 3. `__init__` and `self`

- `__init__` runs automatically when an object is created. It initializes the object.
- `self` refers to the particular object being worked on.
- `self.name` and `self.salary` are **instance attributes**. Each object has its own values.

```python
class Employee:
    def __init__(self, name, salary, department):
        self.name = name
        self.salary = salary
        self.department = department

employee1 = Employee("Ananya", 75000, "Data")
employee2 = Employee("Rahul", 82000, "AI")

print(employee1.name, employee1.salary, employee1.department)
print(employee2.name, employee2.salary, employee2.department)
```

### What does `self` mean?

When you call `employee1.show_details()`, Python passes `employee1` as `self` automatically. You normally do not pass `self` yourself.

## 4. Instance methods

A method is a function defined inside a class. Instance methods use `self` to access that object's data.

```python
class Employee:
    # data
    def __init__(self, name, salary, department):
        self.name = name
        self.salary = salary
        self.department = department
    # --------------------------------
    # transformation applied on top of the data with reference to instance
    def annual_salary(self):
        return self.salary * 12

    def show_details(self):
        print(f"{self.name} works in {self.department}.")
        print(f"Monthly salary: ₹{self.salary:,}")
        print(f"Annual salary: ₹{self.annual_salary():,}")

prudhvi = Employee("PrudhviAkella", 75000, "Data")
prudhvi.show_details()

ravi = Employee("RaviAkella", 82000, "AI")
ravi.show_details()


print(prudhvi.annual_salary())
print(ravi.annual_salary())
```

## 5. A second example: `Customer`

A class can model any real-world concept—not only employees. Here, a customer can place an order and see their total spend.

```python
class Customer:
    def __init__(self, name, email):
        self.name = name
        self.email = email
        self.total_spend = 0

    def place_order(self, amount):
        if amount <= 0:
            print("Order amount must be positive.")
            return
        self.total_spend += amount
        print(f"Order placed for ₹{amount:,}")

    def show_summary(self):
        print(f"Customer: {self.name}")
        print(f"Email: {self.email}")
        print(f"Total spend: ₹{self.total_spend:,}")

customer = Customer("Meera", "meera@example.com")
customer.place_order(1200)
customer.place_order(800)
customer.show_summary()
print("------------------")
customer2 = Customer("Prudhvi", "abc@yahoo.com")
customer2.place_order(1100)
customer2.place_order(700)
customer2.show_summary()
```

## 6. `__repr__`: a useful object representation

`__repr__` controls the readable representation of an object, which is useful while debugging.

```python
def add(a, b):
    return a + b
apythonstring_function = """add(1, 2)"""

eval(apythonstring_function)
```

```python
class Employee:
    def __init__(self, employee_id, name, department):
        self.employee_id = employee_id
        self.name = name
        self.department = department

    def __str__(self):
        return f"Employee ID: {self.employee_id}, Name: {self.name}, department: {self.department}"

    def __repr__(self):
        return f"Employee(employee_id={self.employee_id}, name='{self.name}', department='{self.department}')"

employee1 = Employee(employee_id=101, name='Ananya', department='Data')
employee2 = Employee(employee_id=102, name='Prudhvi Akella', department='Data')
employee3 = Employee(employee_id=103, name='Ravi', department='Data')

employee1_repr = repr(employee1)
employee2_repr = repr(employee2)
employee3_repr = repr(employee3)



with open('employee1.txt', 'w') as file:
    file.write(employee1_repr+'\n')
    file.write(employee2_repr+'\n')
    file.write(employee3_repr+'\n')

# # employee.__str__()
# employee_string = str(employee)
# employee_string
#employee = eval('Employee(101, "Ananya", "Data")')
#print(employee)
```

## 7. `@property`: calculated information

A property lets you access a method like an attribute. Here, `annual_salary` is calculated from monthly salary, so we don't need to store it separately.

```python
class Employee:
    def __init__(self, name, monthly_salary):
        self.name = name
        self.monthly_salary = monthly_salary

    @property
    def annual_salary(self):
        if self.monthly_salary <= 0:
            return 0
        return self.monthly_salary * 12

employee = Employee("Rahul", 82000)
print(employee.annual_salary)  # No parentheses
employee.monthly_salary = 90000
print(employee.annual_salary)  # Recalculated automatically
```

## 8. Work with a list of objects

Once data is represented as objects, you can loop through, filter, and summarize it just like other Python collections.

```python
class Employee:
    def __init__(self, name, department, salary):
        self.name = name
        self.department = department
        self.salary = salary

employees = [
    Employee("Ananya", "Data", 75000),
    Employee("Rahul", "AI", 82000),
    Employee("Meera", "Data", 79000),
]

for employee in employees:
    print(f"{employee.name}: {employee.department}, ₹{employee.salary:,}")

print("\nEmployees in Data:")
for employee in employees:
    if employee.department == "Data":
        print(employee.name)

average_salary = sum(e.salary for e in employees) / len(employees)
print(f"Average monthly salary: ₹{average_salary:,.0f}")
```
