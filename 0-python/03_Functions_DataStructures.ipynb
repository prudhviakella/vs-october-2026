## 1. Functions: write once, reuse anywhere

A **function** is a named block of code. Define it with `def`, then call it by writing its name followed by parentheses.

```python
def greet():
    print("Hello! Welcome.")

greet()
greet()
```

### Parameters and return values

- **Parameters** are the inputs a function receives.
- **`return`** sends a result back to the code that called the function.
- `print()` displays something; `return` lets the caller use the result.

```python
def add(a, b):
    return a + b

result = add(3, 7)
print(result)
print(add(10, 20))
```

## 2. Practical example: calculate a discount

Now let's use the same pattern in a real example:

1. **Input:** product price and discount percentage
2. **Process:** calculate the discount and subtract it from the price
3. **Output:** return the final price

This follows the same input → process → return pattern as the `add()` function above.

```python
def apply_discount(price, discount_percent):
    discount = price * discount_percent / 100
    final_price = price - discount
    return final_price

# Try different prices and discounts
print(apply_discount(1000, 10))  # 10% off 1000 → 900.0
print(apply_discount(2000, 15))  # 15% off 2000 → 1700.0
print(apply_discount(500, 20))   # 20% off 500 → 400.0
```

## 3. Default parameter values

A parameter can have a default value. If the caller leaves it out, Python uses the default. If the caller supplies a value, Python uses that value instead.

This is useful when a function has a common value that works for most calls.

```python
def power(base, exponent=2):
    return base ** exponent

print(power(3))       # Uses exponent=2 → 9
print(power(3, 3))    # Uses exponent=3 → 27
```

## 4. `*args` — accept multiple positional arguments

Sometimes we do not know how many positional values will be passed to a function. Add `*args` to collect extra positional arguments into a **tuple**.

The name `args` is a convention. The `*` symbol is what collects the extra positional arguments.

```python
def rand_function(*numbers):
    print(type(numbers))
    sum = 0
    for number in numbers:
        sum += number
    return sum

print(rand_function(1,2,3,4,5))
```

```python
def add_numbers(*args):
    print("Received:", args)
    return sum(args)

print("Two numbers:", add_numbers(10, 20))
print("Three numbers:", add_numbers(10, 20, 30))
print("Five numbers:", add_numbers(10, 20, 30, 40, 50))
```

**How it works:** Calling `add_numbers(10, 20, 30)` gives the function `args = (10, 20, 30)`. The values are collected into a tuple, and `sum()` adds them.

## 5. `**kwargs` — accept multiple keyword arguments

Sometimes a function needs flexible named settings. Add `**kwargs` to collect extra keyword arguments into a **dictionary**.

The name `kwargs` is a convention. The `**` symbol collects the extra keyword arguments.

```python
def show_config(**kwargs):
    model = kwargs.get("model", "Default Model")
    temperature = kwargs.get("temperature", 0.7)

    print("All settings:", kwargs)
    print(f"Model: {model}")
    print(f"Temperature: {temperature}")

show_config(model="Claude", temperature=0.2)
show_config(model="GPT")
```

**How it works:** Calling `show_config(model="Claude", temperature=0.2)` gives the function a dictionary like `{"model": "Claude", "temperature": 0.2}`. The `.get()` method supplies a default when a setting is missing.

### `*args` vs. `**kwargs`

| Feature | `*args` | `**kwargs` |
|---|---|---|
| Collects | Extra positional arguments | Extra keyword arguments |
| Stored as | Tuple | Dictionary |
| Example call | `func(10, 20)` | `func(model="Claude")` |

### Combining `*args` and `**kwargs`

A function can accept both forms: positional values go into `args`, and named values go into `kwargs`.

```python
def process_request(*args, **kwargs):
    print("Inputs:", args)
    print("Configuration:", kwargs)

process_request(
    "Where is my order?",
    "customer_123",
    model="Claude",
    temperature=0.2
)
```

**Remember:** `args` and `kwargs` are conventional names, not Python keywords. You could use names such as `*numbers` or `**settings`. The `*` and `**` symbols determine how Python collects the arguments.

## 6. Lists: ordered collections

A list stores multiple items in order. Lists are useful when you need to keep a sequence of values and may need to change it.

```python
products = ["Classic Monitor", "Ultimate Perfume", "Yoga Mat Pro"]
print(products)
print(products[0])    # First item
print(len(products))  # Number of items
```

### Common list methods

- `append(x)` — adds one item to the end.
- `extend(items)` — adds each item from another collection.
- `pop()` — removes and returns the last item (or an item at a given index).
- `remove(x)` — removes the first matching value.

**Tip:** `pop()` works by position; `remove()` works by value.

```python
numbers = [3, 1, 4]

numbers.append(2)
print("After append:", numbers)

numbers.extend([5, 9])
print("After extend:", numbers)

last_number = numbers.pop()
print("Removed by pop:", last_number)
print("After pop:", numbers)

numbers.remove(1)
print("After remove(1):", numbers)
```

### A useful application pattern: build message history

Chat applications often store messages as dictionaries inside a list. Each new message is appended to the list.

```python
messages = [
    {"role": "system", "content": "You are a helpful support assistant."}
]

messages.append({"role": "user", "content": "Where is my order?"})
messages.append({"role": "assistant", "content": "Please share your order ID."})

print(messages)
```

## 7. Dictionaries: key-value data

A dictionary stores values under keys. Use it when you want to look up information by a meaningful label, such as a product name or review ID.

```python
product = {"name": "Classic Monitor", "price": 205.21, "stock": 238}

print(product["name"])
print(product["price"])
```

### Accessing values safely

- `data[key]` returns the value, but raises `KeyError` if the key is missing.
- `data.get(key, default)` returns a default value when the key is missing.

```python
product = {"name": "Classic Monitor", "price": 205.21}

print(product["name"])
print(product.get("stock", 0))          # Missing stock → 0
print(product.get("discount", "N/A"))   # Missing discount → N/A
```

### Common dictionary methods

- `keys()` — returns all keys.
- `values()` — returns all values.
- `items()` — returns key-value pairs.
- `get(key, default)` — safely reads a value, with an optional fallback.
- `update(other_dict)` — adds or replaces entries.
- `pop(key)` — removes a key and returns its value.
- `popitem()` — removes and returns the last inserted key-value pair.
- `clear()` — removes all entries.

**Tip:** `pop()` removes a specific key; `popitem()` removes the last inserted pair.

```python
settings = {"model": "example-model", "temperature": 0.2}

print("Keys:", list(settings.keys()))
print("Values:", list(settings.values()))
print("Pairs:", list(settings.items()))
print("Missing setting:", settings.get("region", "not set"))

settings.update({"temperature": 0.0, "max_tokens": 512})
print("After update:", settings)

removed_value = settings.pop("max_tokens")
print("Removed value:", removed_value)
print("After pop:", settings)

removed_pair = settings.popitem()
print("Removed last pair:", removed_pair)
print("After popitem:", settings)

settings.clear()
print("After clear:", settings)
```

### Real use case: index products by name

This creates a dictionary where each product name points to the full product record. After building it, we can look up a product by name.

```python
products = [
    {"name": "Classic Monitor", "price": 205.21, "stock": 238},
    {"name": "Yoga Mat Pro", "price": 45.00, "stock": 150},
]

product_index = {p["name"]: p for p in products}

classic = product_index["Classic Monitor"]
print(f"Price: {classic['price']}  Stock: {classic['stock']}")
```

## 8. Sets: unique values

A set stores unique items. Duplicate values are automatically removed. Sets are useful when you want to compare groups or remove duplicates.

```python
a = {1, 2, 3, 3, 4}
b = {3, 4, 5, 6}

print("a:", a)                     # Duplicate 3 appears only once
print("Union:", a | b)             # Items in either set
print("Intersection:", a & b)      # Items in both sets
print("Difference:", a - b)        # Items in a but not b
```

## 9. Tuples and `zip()`

A **tuple** is an ordered collection that cannot be reassigned item-by-item after creation. It is often used to group related values.

`zip()` pairs items from collections by position.

```python
def min_max(numbers):
    return min(numbers), max(numbers)   # Returns a tuple

result = min_max([3, 1, 9, 4])
print(result)
print("Minimum:", result[0])
print("Maximum:", result[1])
```

```python
names = ["Alice", "Bob", "Carol"]
scores = [90, 75, 88]

for name, score in zip(names, scores):
    print(f"{name}: {score}")
```
