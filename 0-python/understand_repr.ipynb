```python
async def experiment(i):
    # task will go on hold
    await asyncio.sleep(1)
    # task will resume
    print(i)
    return i # task completion

# creation an object of coroutine
coroutine =  experiment(1)
print(coroutine)

result =  asyncio.create_task(coroutine)
# starting task
await result
```

```python
async def experiment(i):
    # task will go on hold
    await asyncio.sleep(1)
    # task will resume
    print(i)
    return i # task completion

exp1 = experiment(1)
exp2 = experiment(2)
exp3 = experiment(3)
await asyncio.gather(exp1, exp2, exp3)
print(30)
```
